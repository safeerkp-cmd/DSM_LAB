```python
import numpy as np


A = np.array([[4, 11],
              [14, 5]], dtype=float)


U, s, Vt = np.linalg.svd(A)


Sigma = np.diag(s)


print("--- Matrix U (Left Singular Vectors) ---")
print(np.round(U, 4))

print("\n--- Matrix Sigma (Singular Values Diagonal) ---")
print(np.round(Sigma, 4))

print("\n--- Matrix Vt (Right Singular Vectors Transposed) ---")
print(np.round(Vt, 4))

A_reconstructed = U @ Sigma @ Vt

print("\n--- Verification (U * Sigma * Vt) ---")
print(np.round(A_reconstructed, 4))
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
iris = load_iris()
X=iris.data
Y=iris.target
print(iris.feature_names)
print(iris.target_names)
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
Knn =KNeighborsClassifier(n_neighbors=3)
Knn.fit(X_train,Y_train)
Y_predict=Knn.predict(X_test)
accuracy=accuracy_score(Y_test,Y_predict)
print("accuracy =" ,accuracy)
print(f'{accuracy:.2f}')

new=[[1.1,2.2,3.5,6.8]]
new_predict =Knn.predict(new)
print(iris.target_names[new_predict])


```
